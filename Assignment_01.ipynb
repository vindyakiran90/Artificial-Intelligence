```python
#1. Create a DataFrame with Random Data
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1, 101, size=(1000, 5)), columns=list('ABCDE'))
df.head()
```

```python
#2. Check for Missing Values
df.isnull().sum()
```

```python
#3. Fill Missing Values with Mean
df.fillna(df.mean(), inplace=True)
df.head()
```

```python
#4. Add a New Column as Sum of Two Columns
df['F'] = df['A'] + df['B']
df.head()
```

```python
#5. Rename Columns
df.rename(columns={'A':'Alpha', 'B':'Beta'}, inplace=True)
df.head()
```

```python
#6. Replace Specific Values
df.replace({'C': {5, 50}}, inplace=True)
df.head()
```

```python
#7. Drop Rows with Any NaN Values
df.dropna(inplace=True)
df.head()
```

```python
#12. Count Unique Values per Group
df = pd.DataFrame({
    'B': ['X', 'X', 'Y', 'Y', 'Y', 'Z', 'Z', 'Z'],
    'C': [10, 20, 10, 10, 30, 20, 20, 40]
})
unique_counts = df.groupby('B')['C'].nunique()
unique_counts
```

```python
#13. Maximum Value per Group
df = pd.DataFrame({
    'A': ['X', 'X', 'Y', 'Y', 'Y', 'Z', 'Z', 'Z'],
    'D': [10, 20, 30, 25, 50, 40, 15, 60]
})
max_values = df.groupby('A')['D'].max()
max_values
```

```python
#14. Apply Custom Function
df = pd.DataFrame({
    'B': ['A', 'A', 'B', 'B', 'C', 'C', 'C'],
    'C': [10, 30, 50, 70, 20, 60, 100]
})
range_per_group = df.groupby('B')['C'].apply(lambda x: x.max() - x.min())
range_per_group
```

```python
#21. Finding Outliers Using IQR
df=pd.DataFrame({'A':[10, 20, 30, 40, 50, 100, 200]})
Q1=df['A'].quantile(0.25)
Q3=df['A'].quantile(0.75)
IQR=Q3-Q1
outliers=df[(df['A']<(Q1-1.5*IQR)) | (df['A']>(Q3+1.5*IQR))]
outliers
```

```python
#37. Calculate Running Total (Cumulative Sum)
data = {'A': [10, 20, 30, 40, 50]}
df = pd.DataFrame(data)
df['Cumulative_Sum'] = df['A'].cumsum()
df.head()
```

```python
#38. Apply a Weighted Mean Function
data = {
    'B': ['X', 'X', 'Y', 'Y', 'Y'],
    'A': [10, 20, 30, 40, 50],
    'C': [2, 3, 1, 4, 5]  # Weights
}
df = pd.DataFrame(data)
df.groupby('B').apply(lambda x: np.average(x['A'], weights=x['C']))
df.head()
```

```python
#40. Grouping by Multiple Columns and Applying Multiple Aggregations
data = {
    'A': ['X', 'X', 'Y', 'Y', 'Z', 'Z'],
    'B': [1, 1, 2, 2, 3, 3],
    'C': [10, 20, 30, 40, 50, 60],
    'D': [5, 15, 25, 35, 45, 55]
}

df = pd.DataFrame(data)
df.groupby(['A', 'B']).agg({'C': ['mean', 'sum'], 'D': ['max', 'min']})
df.head()
```

```python
#41. Creating a Custom Aggregation Function for GroupBy
data = {
    'B': ['X', 'X', 'Y', 'Y', 'Z', 'Z'],
    'A': [10, 20, 30, 40, 50, 60],
    'C': [5, 15, 25, 35, 45, 55]
}

df = pd.DataFrame(data)
df.groupby('B').agg(lambda x: x.median() - x.mean())
df.head()
```

```python
#42. Applying a Custom Sorting Order on a Column
data = {
    'Task': ['Task A', 'Task B', 'Task C', 'Task D'],
    'Priority': ['Medium', 'High', 'Low', 'High']
}

df = pd.DataFrame(data)
custom_order = {'Low': 1, 'Medium': 2, 'High': 3}
df['Priority'] = df['Priority'].map(custom_order)
df.sort_values(by='Priority', inplace=True)
df.head()
```
